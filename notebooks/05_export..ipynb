```python
# Load prepared dataset
prepared = pd.read_csv(
    os.path.join(OUTPUT_DIR, "prepared_city_daily.csv")
)

prepared["Date"] = pd.to_datetime(prepared["Date"])
prepared["Year"] = prepared["Date"].dt.year

# Original pollutant columns
pollutants = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3",
    "Benzene", "Toluene", "Xylene"
]

# Calculate percentage missing by city
city_missingness = (
    prepared.groupby("City")[pollutants]
    .apply(lambda x: x.isna().mean() * 100)
)

# Calculate percentage missing by year
year_missingness = (
    prepared.groupby("Year")[pollutants]
    .apply(lambda x: x.isna().mean() * 100)
)

# Convert to JSON-friendly records
missingness_data = {
    "cities": [
        {"City": city, **values.to_dict()}
        for city, values in city_missingness.iterrows()
    ],
    "years": [
        {"Year": int(year), **values.to_dict()}
        for year, values in year_missingness.iterrows()
    ]
}

# Export
missingness_path = os.path.join(WEBSITE_DIR, "missingness.json")

with open(missingness_path, "w") as f:
    json.dump(missingness_data, f, indent=2)

# Verify
with open(missingness_path, "r") as f:
    check_missingness = json.load(f)

print("Exported:", missingness_path)
print("Cities:", len(check_missingness["cities"]))
print("Years:", len(check_missingness["years"]))

print("\nCity missingness (%):")
display(city_missingness.round(2))

print("\nYear missingness (%):")
display(year_missingness.round(2))
```

```python
# Helper function to load exported JSON files
def load_json(filename):
    path = os.path.join(WEBSITE_DIR, filename)
    with open(path, "r") as f:
        return json.load(f)


# 1. Load previously exported data
metrics = load_json("metrics.json")
classification_curves = load_json("classification_curves.json")
classification_test = load_json("classification_test.json")
example_series = load_json("example_series.json")
test_series = load_json("test_series.json")
unseen_series = load_json("unseen_city.json")
worst_errors = load_json("worst_errors.json")
errors_month = load_json("errors_by_month.json")
errors_city = load_json("errors_by_city.json")
errors_period = load_json("errors_by_period.json")
missingness = load_json("missingness.json")
lasso_importance = load_json("lasso_importance.json")
rf_importance = load_json("rf_importance.json")


# 2. Organize test predictions city-wise
series = {}

for city in sorted(set(row["City"] for row in test_series)):
    city_rows = [
        row for row in test_series if row["City"] == city
    ]

    series[city] = {
        "dates": [row["Target_Date"] for row in city_rows],
        "actual": [row["Actual"] for row in city_rows],
        "period": [row["Period"] for row in city_rows],
        "models": {
            "persistence": [row["Persistence"] for row in city_rows],
            "svr": [row["SVR"] for row in city_rows]
        }
    }


# 3. Organize classification curves
curves = {}

for model in sorted(set(row["Model"] for row in classification_curves)):
    model_rows = sorted(
        [row for row in classification_curves if row["Model"] == model],
        key=lambda x: x["Threshold"]
    )

    model_key = model.lower().replace(" ", "_")

    curves[model_key] = {
        "thresholds": [row["Threshold"] for row in model_rows],
        "precision": [row["Precision"] for row in model_rows],
        "recall": [row["Recall"] for row in model_rows]
    }


# 4. Organize missingness as matrices
pollutants = [
    "PM2.5", "PM10", "NO", "NO2", "NOx", "NH3",
    "CO", "SO2", "O3", "Benzene", "Toluene", "Xylene"
]

missingness_data = {
    "cities": [row["City"] for row in missingness["cities"]],
    "years": [row["Year"] for row in missingness["years"]],
    "pollutants": pollutants,
    "city_values": [
        [row[p] for p in pollutants]
        for row in missingness["cities"]
    ],
    "year_values": [
        [row[p] for p in pollutants]
        for row in missingness["years"]
    ]
}


# 5. Prepare unseen-city metrics
unseen_metrics = pd.read_csv(
    os.path.join(OUTPUT_DIR, "unseen_city_test_metrics_by_city.csv")
)

unseen_results = [
    {
        "city": row["City"],
        "model": "SVR",
        "mae": float(row["MAE"]),
        "rmse": float(row["RMSE"])
    }
    for _, row in unseen_metrics.iterrows()
]


# 6. Prepare worst errors
worst_error_records = [
    {
        "date": row["Target_Date"],
        "city": row["City"],
        "actual": row["PM2.5_next_day"],
        "predicted": row["SVR_prediction"],
        "error": row["Error"],
        "period": row["Period"]
    }
    for row in worst_errors
]


# 7. Prepare final DATA object
DATA = {
    "meta": {
        "cities": sorted(series.keys()),
        "heldOutCities": ["Mumbai", "Guwahati"],
        "threshold": 90,
        "classificationThreshold": 0.40,
        "seed": 42,
        "splits": {
            "train": ["2015-01-02", "2018-12-31"],
            "val": ["2019-01-01", "2019-12-31"],
            "test": ["2020-01-01", "2020-07-01"],
            "lockdownStart": "2020-03-25"
        },
        "models": ["persistence", "svr"]
    },

    "missingness": missingness_data,

    "exampleSeries": example_series,

    "series": series,

    "metrics": metrics,

    "classification": {
        "prauc": {
            "logistic_regression": 0.865191
        },
        "confusion": classification_test["confusion_matrix"],
        "testMetrics": classification_test["metrics"],
        "curves": curves
    },

    "unseen": unseen_results,

    "leakage": {
        "randomCV": {"mae": 20.01446},
        "chronological": {"mae": 24.17621}
    },

    "worstErrors": worst_error_records,

    "importance": {
        "lasso": [
            {
                "feature": row["Feature"],
                "coef": row["Coefficient"]
            }
            for row in lasso_importance
        ],
        "rf": [
            {
                "feature": row["Feature"],
                "importance": row["Importance"]
            }
            for row in rf_importance
        ]
    }
}


# 8. Write JavaScript file
data_js_path = os.path.join(WEBSITE_DIR, "data.js")

with open(data_js_path, "w") as f:
    f.write("const DATA = ")
    json.dump(DATA, f, indent=2, allow_nan=False)
    f.write(";\n")


# 9. Verify the generated file
with open(data_js_path, "r") as f:
    content = f.read()

print("Created:", data_js_path)
print("File size:", round(os.path.getsize(data_js_path) / 1024, 2), "KB")
print("Cities:", DATA["meta"]["cities"])
print("Models:", DATA["meta"]["models"])
print("Test prediction records:", sum(
    len(city["dates"]) for city in DATA["series"].values()
))
print("Classification curve models:", list(DATA["classification"]["curves"].keys()))
print("Unseen cities:", len(DATA["unseen"]))
print("JavaScript declaration found:", content.startswith("const DATA = "))
```

```python
import os
import json
import pandas as pd

OUTPUT_DIR = "/content/drive/MyDrive/PM25_project/outputs"

# --------------------------------------------------
# 1. Helper functions
# --------------------------------------------------

def load_csv(filename):
    path = os.path.join(OUTPUT_DIR, filename)

    if not os.path.exists(path):
        print("Optional file not found:", filename)
        return pd.DataFrame()

    return pd.read_csv(path)


def records(filename):
    df = load_csv(filename)

    if df.empty:
        return []

    # Remove unnecessary exported index columns
    df = df.loc[:, ~df.columns.str.startswith("Unnamed:")]

    # Convert NaN values to JSON null
    return json.loads(
        df.to_json(orient="records", date_format="iso")
    )


def read_text(filename):
    path = os.path.join(OUTPUT_DIR, filename)

    if not os.path.exists(path):
        return ""

    with open(path, "r", encoding="utf-8") as f:
        return f.read()


# --------------------------------------------------
# 2. Build city-wise prediction series
# --------------------------------------------------

regression_predictions = load_csv(
    "final_regression_test_predictions.csv"
)

series = {}

if not regression_predictions.empty:

    for city, group in regression_predictions.groupby("City"):

        group = group.sort_values("Target_Date")

        series[city] = {
            "dates": group["Target_Date"].tolist(),
            "actual": group["PM2.5_next_day"].tolist(),
            "persistence": group["PM2.5"].tolist(),
            "rollingMean": group["PM2.5_roll_mean_7"].tolist(),
            "svr": group["SVR_prediction"].tolist(),
            "period": group["Period"].tolist()
        }


# --------------------------------------------------
# 3. Calculate city-year PM2.5 coverage
# --------------------------------------------------

prepared = load_csv("prepared_city_daily.csv")

city_year_coverage = []

if not prepared.empty:

    prepared["Date"] = pd.to_datetime(prepared["Date"])
    prepared["Year"] = prepared["Date"].dt.year

    coverage = (
        prepared.groupby(["City", "Year"])
        .agg(
            Total_Days=("PM2.5", "size"),
            Observed_Days=("PM2.5", "count")
        )
        .reset_index()
    )

    coverage["Coverage_Percent"] = (
        coverage["Observed_Days"] /
        coverage["Total_Days"] * 100
    )

    city_year_coverage = json.loads(
        coverage.to_json(orient="records")
    )


# --------------------------------------------------
# 4. Calculate pollutant missingness
# --------------------------------------------------

pollutants = [
    "PM2.5", "PM10", "NO", "NO2", "NOx", "NH3",
    "CO", "SO2", "O3", "Benzene", "Toluene", "Xylene"
]

pollutant_missingness = []

if not prepared.empty:

    for pollutant in pollutants:

        if pollutant in prepared.columns:

            missing = int(prepared[pollutant].isna().sum())
            total = len(prepared)

            pollutant_missingness.append({
                "Pollutant": pollutant,
                "Missing": missing,
                "Total": total,
                "Missing_Percent": round(
                    missing / total * 100, 2
                )
            })


# --------------------------------------------------
# 5. Assemble the website data
# --------------------------------------------------

data = {

    "meta": {
        "title": "PM2.5 Prediction and Air Quality Analysis",
        "description": (
            "Next-day PM2.5 regression and poor-air-quality "
            "classification across Indian cities."
        ),
        "pooledCities": [
            "Bengaluru", "Chennai", "Delhi",
            "Hyderabad", "Lucknow", "Patna"
        ],
        "unseenCities": ["Mumbai", "Guwahati"],
        "target": "Next-day PM2.5 concentration",
        "classificationThreshold": 90,
        "testPeriod": "2020-01-01 to 2020-07-01",
        "trainRows": 7751,
        "validationRows": 2188,
        "testRows": 1098,
        "modelReadyRows": 11037
    },

    "dataset": {
        "cityYearCoverage": city_year_coverage,
        "pollutantMissingness": pollutant_missingness
    },

    "series": series,

    "regression": {
        "testMetrics": records("final_test_metrics.csv"),
        "validationBaselines": records("baseline_results_validation.csv"),
        "validationLinearModels": records("linear_models_validation.csv"),
        "validationModelComparison": records("model_comparison_validation.csv"),
        "testPredictions": records("final_regression_test_predictions.csv"),
        "errorsByCity": records("regression_test_errors_by_city.csv"),
        "errorsByMonth": records("regression_test_errors_by_month.csv"),
        "errorsByPeriod": records("regression_test_errors_by_period.csv"),
        "metricsByPeriod": records("regression_test_metrics_by_period.csv"),
        "worstErrors": records("worst_10_test_predictions.csv"),
        "trainingDetails": records("svr_training_details.csv"),
        "chosenModel": read_text("chosen_regression_model.txt"),
        "methods": read_text("regression_methods.md")
    },

    "classification": {
        "testMetrics": records("classification_test_metrics.csv"),
        "testConfusionMatrix": records("classification_test_confusion_matrix.csv"),
        "validationBaselines": records("classification_baseline_validation.csv"),
        "validationConfusionMatrices": records("classification_confusion_matrices_validation.csv"),
        "thresholdCurves": records("classification_threshold_curves.csv"),
        "testPredictions": records("final_classification_test_predictions.csv"),
        "chosenModel": read_text("chosen_classifier.txt")
    },

    "unseen": {
        "testPredictions": records("final_unseen_city_test_predictions.csv"),
        "metricsByCity": records("unseen_city_test_metrics_by_city.csv")
    },

    "interpretation": {
        "randomForestImportance": records("random_forest_feature_importance.csv"),
        "lassoCoefficients": records("lasso_coefficients.csv")
    },

    "documentation": {
        "weeklyExperimentLog": read_text("weekly_experiment_log.md")
    }
}


# --------------------------------------------------
# 6. Export data.js
# --------------------------------------------------

output_path = os.path.join(OUTPUT_DIR, "data.js")

with open(output_path, "w", encoding="utf-8") as f:

    f.write(
        "window.PM25_DATA = "
        + json.dumps(data, indent=2, allow_nan=False)
        + ";\n"
    )


# --------------------------------------------------
# 7. Verify the generated file
# --------------------------------------------------

print("data.js generated successfully.")
print("Location:", output_path)
print("File size:", round(os.path.getsize(output_path) / 1024, 2), "KB")

print("\nIncluded sections:")

for section in data:
    print("-", section)

print("\nCity prediction series:")

for city, city_data in series.items():
    print(city, ":", len(city_data["dates"]), "test observations")

print("\nRegression test prediction rows:", len(regression_predictions))
print(
    "Classification test prediction rows:",
    len(data["classification"]["testPredictions"])
)
```

```python
import os

WEBSITE_DIR = "/content/drive/MyDrive/PM25_project/website"

js_content = r'''
"use strict";

/* =========================================================
   1. DATA AND GLOBAL CONFIGURATION
========================================================= */

const DATA = window.PM25_DATA;

if (!DATA) {
    throw new Error("PM25_DATA was not loaded. Check that data.js is included before app.js.");
}

const COLORS = {
    teal: "#087f69",
    tealLight: "#70e0c0",
    blue: "#719bc5",
    orange: "#d99548",
    red: "#c75d5d",
    purple: "#9683c9",
    grey: "#829095",
    grid: "rgba(120,140,145,0.18)"
};

const charts = {};

Chart.defaults.font.family = '"DM Sans", sans-serif';
Chart.defaults.font.size = 11;
Chart.defaults.color = "#829095";

Chart.defaults.plugins.legend.labels.usePointStyle = true;
Chart.defaults.plugins.legend.labels.boxWidth = 8;

Chart.defaults.scale.grid.color = COLORS.grid;


/* =========================================================
   2. GENERAL HELPER FUNCTIONS
========================================================= */

function normalizeKey(value) {
    return String(value)
        .toLowerCase()
        .replace(/[^a-z0-9]/g, "");
}


function getValue(object, candidates, fallback = null) {

    if (!object || typeof object !== "object") {
        return fallback;
    }

    const keys = Object.keys(object);

    for (const candidate of candidates) {

        const exact = keys.find(
            key => key === candidate
        );

        if (exact !== undefined) {
            return object[exact];
        }

        const normalized = normalizeKey(candidate);

        const match = keys.find(
            key => normalizeKey(key) === normalized
        );

        if (match !== undefined) {
            return object[match];
        }
    }

    return fallback;
}


function toNumber(value) {

    if (value === null || value === undefined || value === "") {
        return null;
    }

    const number = Number(value);

    return Number.isFinite(number) ? number : null;
}


function formatNumber(value, digits = 2) {

    const number = toNumber(value);

    if (number === null) return "—";

    return number.toLocaleString("en-IN", {
        maximumFractionDigits: digits,
        minimumFractionDigits: digits
    });
}


function formatInteger(value) {

    const number = toNumber(value);

    if (number === null) return "—";

    return Math.round(number).toLocaleString("en-IN");
}


function formatDate(value) {

    if (!value) return "—";

    const date = new Date(value);

    if (Number.isNaN(date.getTime())) {
        return String(value);
    }

    return date.toLocaleDateString("en-IN", {
        day: "2-digit",
        month: "short",
        year: "numeric",
        timeZone: "UTC"
    });
}


function getRows(section, key) {

    const value = DATA?.[section]?.[key];

    return Array.isArray(value) ? value : [];
}


function setHTML(id, html) {

    const element = document.getElementById(id);

    if (element) {
        element.innerHTML = html;
    }
}


function createChart(id, config) {

    const canvas = document.getElementById(id);

    if (!canvas) return null;

    if (charts[id]) {
        charts[id].destroy();
    }

    charts[id] = new Chart(canvas, config);

    return charts[id];
}


function emptyMessage(id, message) {
    setHTML(
        id,
        `<p class="empty-state">${message}</p>`
    );
}


function createBarOptions(horizontal = false) {

    return {
        responsive: true,
        maintainAspectRatio: false,
        indexAxis: horizontal ? "y" : "x",
        plugins: {
            legend: {
                position: "bottom"
            }
        },
        scales: {
            x: {
                beginAtZero: true,
                grid: {
                    display: !horizontal
                }
            },
            y: {
                beginAtZero: true,
                grid: {
                    display: horizontal
                }
            }
        }
    };
}


/* =========================================================
   3. MOBILE NAVIGATION
========================================================= */

function initializeNavigation() {

    const toggle = document.getElementById("menuToggle");
    const nav = document.getElementById("mainNav");

    if (!toggle || !nav) return;

    toggle.addEventListener("click", () => {

        const isOpen = nav.classList.toggle("open");

        toggle.setAttribute(
            "aria-expanded",
            String(isOpen)
        );
    });

    nav.querySelectorAll("a").forEach(link => {

        link.addEventListener("click", () => {

            nav.classList.remove("open");

            toggle.setAttribute(
                "aria-expanded",
                "false"
            );
        });
    });
}


/* =========================================================
   4. OVERVIEW STATISTICS
========================================================= */

function renderOverview() {

    const meta = DATA.meta || {};

    const cityCount = Array.isArray(meta.pooledCities)
        ? meta.pooledCities.length
        : 6;

    const testRows = DATA.regression?.testPredictions?.length || 0;

    const unseenRows = DATA.unseen?.testPredictions?.length || 0;

    const values = [
        ["statCities", formatInteger(cityCount)],
        ["statRows", formatInteger(meta.modelReadyRows)],
        ["statTest", formatInteger(testRows)],
        ["statUnseen", formatInteger(unseenRows)]
    ];

    values.forEach(([id, value]) => {
        const element = document.getElementById(id);

        if (element) element.textContent = value;
    });
}


/* =========================================================
   5. HERO CHART
   Uses actual saved observations, not invented data.
========================================================= */

function renderHeroChart() {

    const cities = Object.keys(DATA.series || {});

    if (!cities.length) return;

    const city = cities[0];
    const values = DATA.series[city];

    createChart("heroChart", {
        type: "line",
        data: {
            labels: values.dates,
            datasets: [
                {
                    label: "Actual PM2.5",
                    data: values.actual,
                    borderColor: COLORS.tealLight,
                    backgroundColor: "rgba(112,224,192,0.10)",
                    borderWidth: 1.8,
                    pointRadius: 0,
                    tension: 0.25,
                    fill: true
                },
                {
                    label: "SVR prediction",
                    data: values.svr,
                    borderColor: COLORS.orange,
                    borderWidth: 1.5,
                    pointRadius: 0,
                    tension: 0.25
                }
            ]
        },
        options: {
            responsive: true,
            maintainAspectRatio: false,
            interaction: {
                intersect: false,
                mode: "index"
            },
            plugins: {
                legend: {
                    position: "bottom",
                    labels: {
                        color: "#aab8bb"
                    }
                },
                tooltip: {
                    callbacks: {
                        title: items => formatDate(items[0]?.label)
                    }
                }
            },
            scales: {
                x: {
                    ticks: {
                        color: "#829095",
                        maxTicksLimit: 6
                    },
                    grid: {
                        display: false
                    }
                },
                y: {
                    beginAtZero: true,
                    ticks: {
                        color: "#829095"
                    },
                    grid: {
                        color: "rgba(255,255,255,0.08)"
                    }
                }
            }
        }
    });
}


/* =========================================================
   6. POLLUTANT MISSINGNESS
========================================================= */

function renderMissingness() {

    const rows = DATA.dataset?.pollutantMissingness || [];

    if (!rows.length) {
        emptyMessage(
            "missingnessChart",
            "No pollutant missingness records available."
        );
        return;
    }

    const labels = rows.map(row =>
        getValue(row, ["Pollutant", "feature", "column"])
    );

    const values = rows.map(row =>
        toNumber(
            getValue(row, [
                "Missing_Percent",
                "missing_percent",
                "missing_pct"
            ])
        )
    );

    createChart("missingnessChart", {
        type: "bar",
        data: {
            labels,
            datasets: [{
                label: "Missing values (%)",
                data: values,
                backgroundColor: labels.map(label =>
                    label === "Xylene" || label === "PM10"
                        ? COLORS.orange
                        : COLORS.teal
                ),
                borderRadius: 3,
                barThickness: 15
            }]
        },
        options: {
            ...createBarOptions(true),
            plugins: {
                legend: { display: false },
                tooltip: {
                    callbacks: {
                        label: context =>
                            `${formatNumber(context.raw)}% missing`
                    }
                }
            },
            scales: {
                x: {
                    beginAtZero: true,
                    max: 100,
                    title: {
                        display: true,
                        text: "Missing values (%)"
                    }
                },
                y: {
                    grid: { display: false }
                }
            }
        }
    });
}


/* =========================================================
   7. CITY-YEAR COVERAGE HEATMAP
========================================================= */

function renderCoverage() {

    const rows = DATA.dataset?.cityYearCoverage || [];

    if (!rows.length) {
        emptyMessage(
            "coverageHeatmap",
            "No city-year coverage records available."
        );
        return;
    }

    const cities = [...new Set(
        rows.map(row => getValue(row, ["City"]))
    )].filter(Boolean);

    const years = [...new Set(
        rows.map(row => toNumber(getValue(row, ["Year"])))
    )].filter(value => value !== null).sort((a, b) => a - b);

    function cellColor(value) {

        if (value === null) return "#263640";

        if (value >= 90) return "#167e69";
        if (value >= 70) return "#579b83";
        if (value >= 40) return "#b38a4c";

        return "#a65d58";
    }

    let html = `
        <table class="heatmap-table">
            <thead>
                <tr>
                    <th>City</th>
                    ${years.map(year => `<th>${year}</th>`).join("")}
                </tr>
            </thead>
            <tbody>
    `;

    cities.forEach(city => {

        html += `<tr><td>${city}</td>`;

        years.forEach(year => {

            const record = rows.find(row =>
                getValue(row, ["City"]) === city &&
                toNumber(getValue(row, ["Year"])) === year
            );

            const value = record
                ? toNumber(getValue(record, [
                    "Coverage_Percent",
                    "coverage_percent"
                ]))
                : null;

            const display = value === null
                ? "—"
                : `${formatNumber(value, 1)}%`;

            html += `
                <td
                    style="background:${cellColor(value)};color:#fff"
                    title="${city}, ${year}: ${display}"
                >
                    ${display}
                </td>
            `;
        });

        html += "</tr>";
    });

    html += `
            </tbody>
        </table>
        <p class="footnote">
            Coverage = observed PM2.5 days divided by total daily records.
        </p>
    `;

    setHTML("coverageHeatmap", html);
}


/* =========================================================
   8. REGRESSION VALIDATION COMPARISON
========================================================= */

function renderRegressionComparison() {

    const rows = DATA.regression?.validationModelComparison || [];

    if (!rows.length) {
        emptyMessage(
            "regressionComparisonChart",
            "No validation model comparison records available."
        );
        return;
    }

    const labels = rows.map(row =>
        getValue(row, ["Model", "model"])
    );

    const mae = rows.map(row =>
        toNumber(getValue(row, ["MAE", "mae"]))
    );

    const rmse = rows.map(row =>
        toNumber(getValue(row, ["RMSE", "rmse"]))
    );

    createChart("regressionComparisonChart", {
        type: "bar",
        data: {
            labels,
            datasets: [
                {
                    label: "MAE",
                    data: mae,
                    backgroundColor: COLORS.teal,
                    borderRadius: 3
                },
                {
                    label: "RMSE",
                    data: rmse,
                    backgroundColor: COLORS.blue,
                    borderRadius: 3
                }
            ]
        },
        options: {
            ...createBarOptions(true),
            plugins: {
                legend: {
                    position: "bottom"
                }
            }
        }
    });
}


/* =========================================================
   9. REGRESSION TEST METRICS
========================================================= */

function renderRegressionMetrics() {

    const rows = DATA.regression?.testMetrics || [];

    const preferred = rows.filter(row => {

        const model = String(
            getValue(row, ["Model", "model"], "")
        ).toLowerCase();

        const dataset = String(
            getValue(row, ["Dataset", "Split", "split"], "")
        ).toLowerCase();

        return (
            model.includes("svr") ||
            model.includes("support vector")
        ) && (
            !dataset || dataset.includes("test")
        );
    });

    const row = preferred[0] || rows.find(item =>
        String(getValue(item, ["Model"], "")).toLowerCase().includes("svr")
    );

    if (!row) {
        emptyMessage("regressionMetrics", "Regression metrics unavailable.");
        return;
    }

    const mae = getValue(row, ["MAE", "mae"]);
    const rmse = getValue(row, ["RMSE", "rmse"]);
    const count = getValue(row, ["Rows", "N", "Count"]);

    const cards = [
        ["TEST MAE", mae, "µg/m³"],
        ["TEST RMSE", rmse, "µg/m³"],
        ["TEST OBSERVATIONS", count, "Predictions"]
    ];

    setHTML(
        "regressionMetrics",
        cards.map(([label, value, note]) => `
            <article class="metric-card">
                <span>${label}</span>
                <strong>${label.includes("OBSERVATIONS")
                    ? formatInteger(value)
                    : formatNumber(value, 2)}</strong>
                <small>${note}</small>
            </article>
        `).join("")
    );
}


/* =========================================================
   10. INTERACTIVE REGRESSION EXPLORER
========================================================= */

function renderPredictionExplorer() {

    const citySelect = document.getElementById("citySelect");
    const modelSelect = document.getElementById("modelSelect");
    const periodSelect = document.getElementById("periodSelect");

    if (!citySelect || !modelSelect || !periodSelect) return;

    const cities = Object.keys(DATA.series || {});

    if (!cities.length) return;

    citySelect.innerHTML = cities.map(city =>
        `<option value="${city}">${city}</option>`
    ).join("");

    function update() {

        const city = citySelect.value;
        const model = modelSelect.value;
        const period = periodSelect.value;

        const series = DATA.series[city];

        if (!series) return;

        const predictionKey = model;

        const indices = series.dates.map((date, index) => index)
            .filter(index =>
                period === "all" ||
                series.period[index] === period
            );

        const labels = indices.map(index => series.dates[index]);

        const actual = indices.map(index => series.actual[index]);

        const predicted = indices.map(index =>
            series[predictionKey]?.[index] ?? null
        );

        createChart("predictionChart", {
            type: "line",
            data: {
                labels,
                datasets: [
                    {
                        label: "Actual PM2.5",
                        data: actual,
                        borderColor: COLORS.teal,
                        backgroundColor: "rgba(8,127,105,0.08)",
                        borderWidth: 2,
                        pointRadius: 0,
                        tension: 0.2
                    },
                    {
                        label: modelSelect.options[
                            modelSelect.selectedIndex
                        ].text,
                        data: predicted,
                        borderColor: COLORS.orange,
                        borderWidth: 1.7,
                        pointRadius: 0,
                        tension: 0.2
                    }
                ]
            },
            options: {
                responsive: true,
                maintainAspectRatio: false,
                interaction: {
                    intersect: false,
                    mode: "index"
                },
                plugins: {
                    legend: {
                        position: "bottom"
                    },
                    tooltip: {
                        callbacks: {
                            title: items => formatDate(items[0]?.label)
                        }
                    }
                },
                scales: {
                    x: {
                        ticks: {
                            maxTicksLimit: 9
                        },
                        grid: {
                            display: false
                        }
                    },
                    y: {
                        beginAtZero: true,
                        title: {
                            display: true,
                            text: "PM2.5 (µg/m³)"
                        }
                    }
                }
            }
        });
    }

    [citySelect, modelSelect, periodSelect].forEach(control => {
        control.addEventListener("change", update);
    });

    update();
}


/* =========================================================
   11. CLASSIFICATION METRICS
========================================================= */

function renderClassificationMetrics() {

    const rows = DATA.classification?.testMetrics || [];

    const row = rows[0];

    if (!row) {
        emptyMessage(
            "classificationMetrics",
            "Classification test metrics unavailable."
        );
        return;
    }

    const metrics = [
        ["PRECISION", ["Precision", "precision"]],
        ["RECALL", ["Recall", "recall"]],
        ["F1 SCORE", ["F1", "F1_score", "F1-score"]],
        ["PR-AUC", ["PR_AUC", "PRAUC", "Average_Precision"]],
        ["ACCURACY", ["Accuracy", "accuracy"]]
    ];

    setHTML(
        "classificationMetrics",
        metrics.map(([label, keys]) => {

            const value = getValue(row, keys);

            return `
                <article class="metric-card">
                    <span>${label}</span>
                    <strong>${formatNumber(value, 3)}</strong>
                    <small>Held-out test evaluation</small>
                </article>
            `;
        }).join("")
    );
}


/* =========================================================
   12. CONFUSION MATRIX
========================================================= */

function renderConfusionMatrix() {

    const rows = DATA.classification?.testConfusionMatrix || [];

    if (!rows.length) {
        emptyMessage(
            "confusionMatrix",
            "Confusion matrix data unavailable."
        );
        return;
    }

    let tn = null;
    let fp = null;
    let fn = null;
    let tp = null;

    const first = rows[0];

    tn = getValue(first, ["TN", "True_Negative"]);
    fp = getValue(first, ["FP", "False_Positive"]);
    fn = getValue(first, ["FN", "False_Negative"]);
    tp = getValue(first, ["TP", "True_Positive"]);

    if ([tn, fp, fn, tp].some(value => value === null)) {

        if (rows.length >= 2) {

            const numericKeys = Object.keys(rows[0]).filter(key =>
                toNumber(rows[0][key]) !== null
            );

            if (numericKeys.length >= 2) {

                const values = rows.map(row =>
                    numericKeys.slice(0, 2).map(key =>
                        toNumber(row[key])
                    )
                );

                tn = values[0][0];
                fp = values[0][1];
                fn = values[1][0];
                tp = values[1][1];
            }
        }
    }

    if ([tn, fp, fn, tp].some(value => value === null)) {
        emptyMessage(
            "confusionMatrix",
            "Could not identify the confusion-matrix values from the exported columns."
        );
        return;
    }

    setHTML("confusionMatrix", `
        <table class="confusion-table">
            <thead>
                <tr>
                    <th></th>
                    <th colspan="2">PREDICTED</th>
                </tr>
                <tr>
                    <th>ACTUAL</th>
                    <th>Negative</th>
                    <th>Positive</th>
                </tr>
            </thead>
            <tbody>
                <tr>
                    <th>Negative</th>
                    <td><strong>${formatInteger(tn)}</strong><span>True negative</span></td>
                    <td><strong>${formatInteger(fp)}</strong><span>False positive</span></td>
                </tr>
                <tr>
                    <th>Positive</th>
                    <td><strong>${formatInteger(fn)}</strong><span>False negative</span></td>
                    <td><strong>${formatInteger(tp)}</strong><span>True positive</span></td>
                </tr>
            </tbody>
        </table>
    `);
}


/* =========================================================
   13. CLASSIFICATION THRESHOLD CURVE
========================================================= */

function renderThresholdChart() {

    const rows = DATA.classification?.thresholdCurves || [];

    if (!rows.length) {
        emptyMessage(
            "thresholdChart",
            "Classification threshold records unavailable."
        );
        return;
    }

    const models = [...new Set(rows.map(row =>
        getValue(row, ["Model", "model"])
    ))].filter(Boolean);

    const datasets = [];

    models.forEach((model, index) => {

        const modelRows = rows.filter(row =>
            getValue(row, ["Model", "model"]) === model
        ).sort((a, b) =>
            toNumber(getValue(a, ["Threshold", "threshold"])) -
            toNumber(getValue(b, ["Threshold", "threshold"]))
        );

        datasets.push({
            label: `${model} — Precision`,
            data: modelRows.map(row => ({
                x: toNumber(getValue(row, ["Threshold", "threshold"])),
                y: toNumber(getValue(row, ["Precision", "precision"]))
            })),
            borderColor: index === 0 ? COLORS.teal : COLORS.blue,
            backgroundColor: "transparent",
            borderWidth: 2,
            pointRadius: 3,
            tension: 0.2
        });

        datasets.push({
            label: `${model} — Recall`,
            data: modelRows.map(row => ({
                x: toNumber(getValue(row, ["Threshold", "threshold"])),
                y: toNumber(getValue(row, ["Recall", "recall"]))
            })),
            borderColor: index === 0 ? COLORS.orange : COLORS.purple,
            backgroundColor: "transparent",
            borderWidth: 1.8,
            borderDash: [5, 4],
            pointRadius: 3,
            tension: 0.2
        });
    });

    createChart("thresholdChart", {
        type: "line",
        data: { datasets },
        options: {
            responsive: true,
            maintainAspectRatio: false,
            plugins: {
                legend: {
                    position: "bottom"
                }
            },
            scales: {
                x: {
                    type: "linear",
                    min: 0,
                    max: 1,
                    title: {
                        display: true,
                        text: "Classification threshold"
                    }
                },
                y: {
                    min: 0,
                    max: 1,
                    title: {
                        display: true,
                        text: "Score"
                    }
                }
            }
        }
    });
}


/* =========================================================
   14. CITY-WISE ERROR CHART
========================================================= */

function renderCityErrors() {

    const rows = DATA.regression?.errorsByCity || [];

    if (!rows.length) {
        emptyMessage(
            "cityErrorChart",
            "City-wise error records unavailable."
        );
        return;
    }

    const labels = rows.map(row =>
        getValue(row, ["City", "city"])
    );

    const values = rows.map(row =>
        toNumber(getValue(row, ["MAE", "mae"]))
    );

    createChart("cityErrorChart", {
        type: "bar",
        data: {
            labels,
            datasets: [{
                label: "MAE (µg/m³)",
                data: values,
                backgroundColor: COLORS.teal,
                borderRadius: 3
            }]
        },
        options: createBarOptions(true)
    });
}


/* =========================================================
   15. MONTHLY ERROR CHART
========================================================= */

function renderMonthlyErrors() {

    const rows = DATA.regression?.errorsByMonth || [];

    if (!rows.length) {
        emptyMessage(
            "monthlyErrorChart",
            "Monthly error records unavailable."
        );
        return;
    }

    const labels = rows.map(row =>
        getValue(row, ["Month", "month", "YearMonth", "year_month"])
    );

    const values = rows.map(row =>
        toNumber(getValue(row, ["MAE", "mae"]))
    );

    createChart("monthlyErrorChart", {
        type: "line",
        data: {
            labels,
            datasets: [{
                label: "MAE (µg/m³)",
                data: values,
                borderColor: COLORS.teal,
                backgroundColor: "rgba(8,127,105,0.10)",
                fill: true,
                tension: 0.25,
                pointRadius: 3
            }]
        },
        options: {
            responsive: true,
            maintainAspectRatio: false,
            plugins: {
                legend: {
                    display: false
                }
            },
            scales: {
                x: {
                    ticks: {
                        maxTicksLimit: 8
                    }
                },
                y: {
                    beginAtZero: true
                }
            }
        }
    });
}


/* =========================================================
   16. PERIOD COMPARISON
========================================================= */

function renderPeriodChart() {

    const rows = DATA.regression?.metricsByPeriod || [];

    if (!rows.length) {
        emptyMessage(
            "periodChart",
            "Period-wise regression metrics unavailable."
        );
        return;
    }

    const models = [...new Set(rows.map(row =>
        getValue(row, ["Model", "model"])
    ))].filter(Boolean);

    const periods = [...new Set(rows.map(row =>
        getValue(row, ["Period", "period"])
    ))].filter(Boolean);

    const datasets = models.map((model, index) => {

        const color = [
            COLORS.teal,
            COLORS.orange,
            COLORS.blue
        ][index % 3];

        return {
            label: model,
            data: periods.map(period => {

                const row = rows.find(item =>
                    getValue(item, ["Model", "model"]) === model &&
                    getValue(item, ["Period", "period"]) === period
                );

                return row
                    ? toNumber(getValue(row, ["MAE", "mae"]))
                    : null;
            }),
            backgroundColor: color,
            borderRadius: 3
        };
    });

    createChart("periodChart", {
        type: "bar",
        data: {
            labels: periods,
            datasets
        },
        options: {
            ...createBarOptions(false),
            plugins: {
                legend: {
                    position: "bottom"
                }
            }
        }
    });
}


/* =========================================================
   17. WORST PREDICTIONS TABLE
========================================================= */

function renderWorstErrors() {

    const rows = DATA.regression?.worstErrors || [];

    if (!rows.length) {
        setHTML(
            "worstErrorsTable",
            '<tr><td colspan="6">No worst-error records available.</td></tr>'
        );
        return;
    }

    const html = rows.map(row => {

        const date = getValue(row, ["date", "Target_Date", "Date"]);
        const city = getValue(row, ["city", "City"]);
        const actual = getValue(row, ["actual", "Actual"]);
        const predicted = getValue(row, ["predicted", "Prediction"]);
        const error = getValue(row, ["error", "Absolute_Error", "Abs_Error"]);
        const period = getValue(row, ["period", "Period"]);

        return `
            <tr>
                <td>${formatDate(date)}</td>
                <td>${city ?? "—"}</td>
                <td>${formatNumber(actual)}</td>
                <td>${formatNumber(predicted)}</td>
                <td>${formatNumber(error)}</td>
                <td>${period ?? "—"}</td>
            </tr>
        `;
    }).join("");

    setHTML("worstErrorsTable", html);
}


/* =========================================================
   18. UNSEEN CITY EVALUATION
========================================================= */

function renderUnseenCities() {

    const rows = DATA.unseen?.metricsByCity || [];

    if (!rows.length) {
        emptyMessage(
            "unseenCityMetrics",
            "Unseen-city metrics unavailable."
        );
        return;
    }

    const html = rows.map(row => {

        const city = getValue(row, ["City", "city"], "Unknown");
        const mae = getValue(row, ["MAE", "mae"]);
        const rmse = getValue(row, ["RMSE", "rmse"]);
        const count = getValue(row, ["Rows", "Count", "N"]);

        return `
            <article class="unseen-card">
                <h3>${city}</h3>

                <div class="unseen-stat">
                    <span>MAE</span>
                    <strong>${formatNumber(mae)} µg/m³</strong>
                </div>

                <div class="unseen-stat">
                    <span>RMSE</span>
                    <strong>${formatNumber(rmse)} µg/m³</strong>
                </div>

                <div class="unseen-stat">
                    <span>Observations</span>
                    <strong>${formatInteger(count)}</strong>
                </div>
            </article>
        `;
    }).join("");

    setHTML("unseenCityMetrics", html);

    createChart("unseenChart", {
        type: "bar",
        data: {
            labels: rows.map(row =>
                getValue(row, ["City", "city"])
            ),
            datasets: [
                {
                    label: "SVR MAE",
                    data: rows.map(row =>
                        toNumber(getValue(row, ["MAE", "mae"]))
                    ),
                    backgroundColor: COLORS.teal,
                    borderRadius: 3
                },
                {
                    label: "SVR RMSE",
                    data: rows.map(row =>
                        toNumber(getValue(row, ["RMSE", "rmse"]))
                    ),
                    backgroundColor: COLORS.orange,
                    borderRadius: 3
                }
            ]
        },
        options: createBarOptions(false)
    });
}


/* =========================================================
   19. RANDOM FOREST FEATURE IMPORTANCE
========================================================= */

function renderImportance() {

    const rows = DATA.interpretation?.randomForestImportance || [];

    if (!rows.length) {
        emptyMessage(
            "importanceChart",
            "Feature-importance records unavailable."
        );
        return;
    }

    const featureKey = Object.keys(rows[0]).find(key =>
        ["feature", "feature_name", "variable"].includes(
            normalizeKey(key)
        )
    );

    const importanceKey = Object.keys(rows[0]).find(key =>
        ["importance", "feature_importance", "importance_score"].includes(
            normalizeKey(key)
        )
    );

    if (!featureKey || !importanceKey) {
        emptyMessage(
            "importanceChart",
            "Could not identify feature and importance columns."
        );
        return;
    }

    const sorted = [...rows].sort((a, b) =>
        Number(b[importanceKey]) - Number(a[importanceKey])
    );

    createChart("importanceChart", {
        type: "bar",
        data: {
            labels: sorted.map(row => row[featureKey]),
            datasets: [{
                label: "Feature importance",
                data: sorted.map(row =>
                    toNumber(row[importanceKey])
                ),
                backgroundColor: COLORS.teal,
                borderRadius: 3
            }]
        },
        options: createBarOptions(true)
    });
}


/* =========================================================
   20. LASSO COEFFICIENTS
========================================================= */

function renderLasso() {

    const rows = DATA.interpretation?.lassoCoefficients || [];

    if (!rows.length) {
        emptyMessage(
            "lassoChart",
            "Lasso coefficient records unavailable."
        );
        return;
    }

    const featureKey = Object.keys(rows[0]).find(key =>
        ["feature", "feature_name", "variable"].includes(
            normalizeKey(key)
        )
    );

    const coefficientKey = Object.keys(rows[0]).find(key =>
        ["coefficient", "coef", "lasso_coefficient"].includes(
            normalizeKey(key)
        )
    );

    if (!featureKey || !coefficientKey) {
        emptyMessage(
            "lassoChart",
            "Could not identify feature and coefficient columns."
        );
        return;
    }

    const sorted = [...rows].sort((a, b) =>
        Math.abs(Number(b[coefficientKey])) -
        Math.abs(Number(a[coefficientKey]))
    );

    createChart("lassoChart", {
        type: "bar",
        data: {
            labels: sorted.map(row => row[featureKey]),
            datasets: [{
                label: "Lasso coefficient",
                data: sorted.map(row =>
                    toNumber(row[coefficientKey])
                ),
                backgroundColor: sorted.map(row =>
                    Number(row[coefficientKey]) >= 0
                        ? COLORS.teal
                        : COLORS.orange
                ),
                borderRadius: 3
            }]
        },
        options: {
            ...createBarOptions(true),
            scales: {
                x: {
                    grid: {
                        color: COLORS.grid
                    }
                },
                y: {
                    grid: {
                        display: false
                    }
                }
            }
        }
    });
}


/* =========================================================
   21. INITIALIZE ALL WEBSITE COMPONENTS
========================================================= */

function initializeWebsite() {

    initializeNavigation();

    renderOverview();

    renderHeroChart();

    renderMissingness();

    renderCoverage();

    renderRegressionMetrics();

    renderRegressionComparison();

    renderPredictionExplorer();

    renderClassificationMetrics();

    renderConfusionMatrix();

    renderThresholdChart();

    renderCityErrors();

    renderMonthlyErrors();

    renderPeriodChart();

    renderWorstErrors();

    renderUnseenCities();

    renderImportance();

    renderLasso();

    console.log("PM2.5 research website initialized.");
    console.log("Available project data:", DATA);
}


if (document.readyState === "loading") {

    document.addEventListener(
        "DOMContentLoaded",
        initializeWebsite
    );

} else {

    initializeWebsite();
}
'''

output_path = os.path.join(WEBSITE_DIR, "app.js")

with open(output_path, "w", encoding="utf-8") as f:
    f.write(js_content)

print("app.js created:", os.path.exists(output_path))
print("File size:", round(os.path.getsize(output_path) / 1024, 2), "KB")

print("\nWebsite folder contents:")
for filename in sorted(os.listdir(WEBSITE_DIR)):
    print("-", filename)
```
